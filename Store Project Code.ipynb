### 4 : Arithmethic Operation

```python
final_amount = df["Amount"].sum()
print("Total Amount :",final_amount)
```

```python
# Central Tendency of Amount

mean = df["Amount"].mean()
median = df["Amount"].median()
mode = df["Amount"].mode()[0]
lar = df["Amount"].max()
small = df["Amount"].min()
std = df["Amount"].std()
skew = df["Amount"].skew().round(2)

print("Central Tendency of Amount \n")
print("Mean :",mean)
print("Median :",median)
print("Mode :",mode)
print("Standard Deviation :",std)
print("Maximum :",lar)
print("Minimum :",small)
print("Skewness :",skew)
```

```python
# Central Tendency of Age

mean = df["Age"].mean()
median = df["Age"].median()
mode = df["Age"].mode()[0]
lar = df["Age"].max()
small = df["Age"].min()
std = df["Age"].std()
skew = df["Age"].skew().round(2)

print("Central Tendency of Age \n")
print("Mean :",mean)
print("Median :",median)
print("Mode :",mode)
print("Standard Deviation :",std)
print("Maximum :",lar)
print("Minimum :",small)
print("Skewness :",skew)
```

### 4 : Adding Column

```python
df["Age_Group"] = df["Age"].apply(lambda x: "Teen" if 18 <= x <= 20 
                                                   else "Adult" if 21 <= x <= 30 
                                                   else "Old")
print(df)
```

```python
# Pivot Table Top 7 cities for shiping

x=pd.pivot_table(
    df,
    values = "Order Id",
    index = "Ship-State",
    aggfunc = "count"
)

top7 = x.sort_values(by="Order Id", ascending=False).head(7)

print(top7)
```

```python
plt.plot(top7.index, top7.values,marker = "o")
plt.title("Top 7 Ship States by Orders")
plt.xlabel("State")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45)
plt.show()
```
